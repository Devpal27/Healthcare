# Diseases Analysis

```python
pd.read_sql_query("""select "Medical Condition",
Gender, Age,
Medication from healthcare Where Age = (select Age from healthcare where Age < 18 as Childeren;""",conn)
```

```python
pd.read_sql_query("""
SELECT
    CASE
        WHEN Age BETWEEN 0 AND 17 THEN '0-17'
        WHEN Age BETWEEN 18 AND 60 THEN '18-59'
        WHEN Age BETWEEN 60 AND 89 THEN '59-89'
    END AS age_group,

    Gender,
    "Medical Condition",
    Medication,
    COUNT(*) AS patient_count

FROM healthcare
WHERE Age < 90

GROUP BY
    age_group,
    Gender,
    "Medical Condition",
    Medication

ORDER BY patient_count DESC;
""", conn)
```
